```python
with open("bandung_drive_osm.pkl", "rb") as f:
    G = pickle.load(f)
```

```python
# Ambil semua edge dalam bentuk generator
edges = G.edges(data=True)

# Ubah ke list agar bisa diakses
edges_list = list(edges)

# Cetak 5 edge pertama beserta semua field-nya
for i, (u, v, data) in enumerate(edges_list[:5]):
    print(f"Edge {i+1}:")
    print(f"  From node: {u}")
    print(f"  To node:   {v}")
    print(f"  Attributes:")
    for key, value in data.items():
        print(f"    {key}: {value}")
    print()
```
